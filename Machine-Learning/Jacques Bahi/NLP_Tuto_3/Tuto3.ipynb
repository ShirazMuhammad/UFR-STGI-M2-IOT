```python
# Task 1: Create an nlp doc from Completetext.txt and name it complete_doc
import spacy
from wasabi import Printer

# Load the English medium spaCy model
nlp = spacy.load("en_core_web_md")

# Read text content from Completetext.txt
with open("completetext.txt", "r", encoding="utf-8") as file:
    text_data = file.read()

# Generate the spaCy doc object
complete_doc = nlp(text_data)

print("=== TASK 1: COMPLETE_DOC CREATED ===")
print("Doc Character Count:", len(complete_doc.text))
```

```python
# Task 2: Extract all the tokens
print("=== TASK 2: ALL TOKENS EXTRACTION ===")

# Extract all tokens into a list
all_tokens = [token.text for token in complete_doc]

print(f"Total Tokens Extracted: {len(all_tokens)}\n")
print("Tokens List:")
for token in complete_doc:
    print(token.text)
```

```python
# Task 3: Print all Lemmas in complete_doc
print("=== TASK 3: TOKENS & LEMMAS ===")
print(f"{'TOKEN':<20} | {'LEMMA':<20}")
print("-" * 45)

for token in complete_doc:
    if not token.is_space: # Filtering newlines and spaces
        print(f"{token.text:<20} | {token.lemma_:<20}")
```

```python
# Task 4: Remove stop words then compute the words frequencies
from collections import Counter

print("=== TASK 4: FILTER STOP WORDS & COMPUTE FREQUENCIES ===")

# Filter out stop words, punctuation marks, spaces, and numbers
filtered_tokens = [
    token.text.lower()
    for token in complete_doc
    if not token.is_stop and not token.is_punct and not token.is_space and token.is_alpha
]

# Calculate word frequencies
word_frequencies = Counter(filtered_tokens)

print(f"Filtered Token Count (no stop words): {len(filtered_tokens)}")
print(f"Total Word Frequency Keys: {len(word_frequencies)}")
```

```python
# Task 5: Print the 5 commonly occurring words with their frequencies
print("=== TASK 5: TOP 5 COMMONLY OCCURRING WORDS ===")
print(f"{'WORD':<20} | {'FREQUENCY':<10}")
print("-" * 35)

top_5_words = word_frequencies.most_common(5)
for word, freq in top_5_words:
    print(f"{word:<20} | {freq:<10}")
```

```python
# Task 6: Print the unique words
print("=== TASK 6: UNIQUE WORDS ===")

# Create a sorted list of unique words
unique_words = sorted(list(set(filtered_tokens)))

print(f"Total Unique Words: {len(unique_words)}\n")
for word in unique_words:
    print(word)
```

```python
# Task 7: Print tokens, token.tag, token.pos, and spacy explanation of token.tag
print("=== TASK 7: PART OF SPEECH TAGS & EXPLANATIONS ===")
print(f"{'TOKEN':<20} | {'POS':<10} | {'TAG':<8} | {'TAG EXPLANATION'}")
print("-" * 80)

for token in complete_doc:
    if not token.is_space:
        tag_explanation = spacy.explain(token.tag_)
        print(f"{token.text:<20} | {token.pos_:<10} | {token.tag_:<8} | {tag_explanation}")
```

```python
# Task 8: Extract the Nouns and the Adjectives
print("=== TASK 8: EXTRACT NOUNS & ADJECTIVES ===")

# Extract unique Nouns (NOUN and PROPN)
nouns = sorted(list(set([token.text.lower() for token in complete_doc if token.pos_ in ["NOUN", "PROPN"]])))

# Extract unique Adjectives (ADJ)
adjectives = sorted(list(set([token.text.lower() for token in complete_doc if token.pos_ == "ADJ"])))

print(f"--- Extracted Nouns ({len(nouns)} total) ---")
print(nouns)

print(f"\n--- Extracted Adjectives ({len(adjectives)} total) ---")
print(adjectives)
```
