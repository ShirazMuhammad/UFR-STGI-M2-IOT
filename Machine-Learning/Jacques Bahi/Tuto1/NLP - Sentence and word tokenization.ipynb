```python
import spacy

# 1 & 2: Load spaCy and the English small package
nlp = spacy.load('en_core_web_sm')

# 3: Create the doc object with the provided text[cite: 2]
text = "Pr. Makhoul explains a reserach project. His students are listening to him"
doc = nlp(text)

# 4: Split the paragraph into sentences and print them[cite: 2]
print("--- Sentences ---")
for sent in doc.sents:
    print(sent.text)

# 5: Print all individual words (tokens) in each sentence[cite: 2]
print("\n--- Words in each sentence ---")
for sent in doc.sents:
    print(f"Sentence: {sent.text}")
    for token in sent:
        print(f"  Word: {token.text}")
```

```python
import spacy

# Load the spaCy model
nlp = spacy.load('en_core_web_md')

# Provided text about Gus Proto
text = (
    "Gus Proto is a Python developer currently working for a London-based Fintech company. "
    "He is interested in learning Natural Language Processing. There is a developer conference "
    "happening on 21 July 2019 in London. It is titled \"Applications of Natural Language Processing\". "
    "There is a helpline number available at +1-1234567891. Gus is helping organize it. He keeps "
    "organizing local Python meetups and several internal talks at his workplace. Gus is also presenting "
    "a talk. The talk will introduce the reader about \"Use cases of Natural Language Processing in Fintech\". "
    "Apart from his work, he is very passionate about music. Gus is learning to play the Piano. He has "
    "enrolled himself in the weekend batch of Great Piano Academy. Great Piano Academy is situated in "
    "Mayfair or the City of London and has worldclass piano instructors."
)

doc = nlp(text)

print("--- Filtered Tokens & Lowercase Lemmas ---")
for token in doc:
    # Exclude stop words and punctuation marks
    if not token.is_stop and not token.is_punct:
        print(f"Original: {token.text:<20} | Lowercase Lemma: {token.lemma_.lower()}")
```
