```python
# Task 1: Visualize and Save a Dependency Tree as an SVG
import spacy
from spacy import displacy
from pathlib import Path

# 1. Load the spaCy model
nlp = spacy.load('en_core_web_md')

# 2. Parse an arbitrary string into a doc object
text = "The quick brown fox jumps over the lazy dog."
doc = nlp(text)

# 3. Generate the dependency tree as SVG markup (jupyter=False prevents notebook blocking)
svg = displacy.render(doc, style='dep', jupyter=False)

# 4. Define the file path and save the SVG graphic to your computer
output_path = Path("dependency_plot.svg")
output_path.open("w", encoding="utf-8").write(svg)

print("Dependency tree successfully saved as 'dependency_plot.svg'!")
```

```python
# Task 2: Document Merging and Text Spans
import spacy

# Load the spaCy English medium model
nlp = spacy.load('en_core_web_md')

# 1. Define custom strings customized to your university and field
string1 = "Mastering Natural Language Processing is crucial for modern IoT architectures."
string2 = "Students at Université Marie et Louis Pasteur develop sophisticated AI solutions."

doc1 = nlp(string1)
doc2 = nlp(string2)

# 2. Show texts and individual tokens with a clean layout
print("-" * 60)
print("TASK 2a: INDIVIDUAL STRING PARSING")
print("-" * 60)
print(f"📄 String 1: {doc1.text}")
print(f"🔠 Tokens:   {[token.text for token in doc1]}\n")

print(f"📄 String 2: {doc2.text}")
print(f"🔠 Tokens:   {[token.text for token in doc2]}\n")

# 3. Merge the two strings together and parse the new combined text
merged_text = string1 + " " + string2
merged_doc = nlp(merged_text)

# 4. Show the sentences in the new merged doc
print("-" * 60)
print("TASK 2b: DOCUMENT MERGING & SENTENCE EXTRACTION")
print("-" * 60)
for i, sent in enumerate(merged_doc.sents, 1):
    print(f"Sentence {i}: {sent.text}")

# 5. Print tokens and their base lemmas in a formatted table
print("\n" + "-" * 60)
print("TASK 2c: TOKENIZATION & LEMMATIZATION (Punctuation Filtered)")
print("-" * 60)
# Added a Part-of-Speech (POS) column for an extra professional touch
print(f"{'TOKEN':<20} | {'LEMMA':<20} | {'PART OF SPEECH'}")
print("-" * 60)
for token in merged_doc:
    if not token.is_punct:
        print(f"{token.text:<20} | {token.lemma_:<20} | {token.pos_}")

# 6 & 7. Test the span function and print tokens for the span
# Slicing from index 0 to 6 captures "Mastering Natural Language Processing is crucial"
sample_span = merged_doc[0:6]

print("\n" + "-" * 60)
print("TASK 2d: SPAN EXTRACTION (Indices 0 to 6)")
print("-" * 60)
print(f"Extracted Span: '{sample_span.text}'\n")
print("Tokens within this span:")
for i, token in enumerate(sample_span):
    print(f"  {i}. {token.text}")
print("-" * 60)
```

```python
### Task 3: Pipeline Modification & Custom Boundaries
import spacy
from spacy.language import Language

# Initialize the spaCy medium model
nlp = spacy.load('en_core_web_md')

# The mandatory string provided in the assignment guidelines
target_text = "Jean-Claude, can you help please?,..., I will give it you back! Never forget... Thank you"

# =====================================================================
# TASK 3a: PARSE TEXT, PRINT TOKENS, LEMMAS, & DEFAULT SENTENCES
# =====================================================================
baseline_doc = nlp(target_text)

print("=" * 70)
print("TASK 3a: TOKENS & LEMMAS (BASELINE PIPELINE)")
print("=" * 70)
print(f"{'RAW TOKEN':<15} | {'ROOT LEMMA':<15} | {'PART OF SPEECH'}")
print("-" * 70)
for token in baseline_doc:
    print(f"{token.text:<15} | {token.lemma_:<15} | {token.pos_}")


# =====================================================================
# TASK 3b: CREATE & INJECT NEW CUSTOM DELIMITER USING '...'
# =====================================================================
@Language.component("ellipsis_boundary_interceptor")
def apply_ellipsis_boundaries(doc):
    # Iterate through tokens up to the second-to-last token
    for i, token in enumerate(doc[:-1]):
        if "..." in token.text:
            # Force the subsequent token to start a new sentence
            doc[i + 1].is_sent_start = True
    return doc

print("\n" + "=" * 70)
print("TASK 3b: INJECTING CUSTOM DELIMITER COMPONENT")
print("=" * 70)
if "ellipsis_boundary_interceptor" not in nlp.pipe_names:
    nlp.add_pipe("ellipsis_boundary_interceptor", before="parser")
    print(f"Status: Component successfully injected.")
    print(f"Active Pipeline Components: {nlp.pipe_names}")
else:
    print(f"Status: Component already active.")
print("=" * 70)


# =====================================================================
# TASK 3c: DETECT SENTENCES WITH AND WITHOUT CUSTOMIZATION
# =====================================================================
print("\n" + "=" * 70)
print("TASK 3c (Part 1): SENTENCES WITHOUT CUSTOMIZATION")
print("=" * 70)
for idx, sent in enumerate(baseline_doc.sents, 1):
    print(f"  [Sentence {idx}] {sent.text}")

# Reprocess the text through the upgraded pipeline with the custom delimiter rule
custom_doc = nlp(target_text)

print("\n" + "=" * 70)
print("TASK 3c (Part 2): SENTENCES WITH CUSTOMIZATION ('...' AS DELIMITER)")
print("=" * 70)
for idx, sent in enumerate(custom_doc.sents, 1):
    print(f"  [Sentence {idx}] {sent.text}")
print("=" * 70)
```

```python
# Task 4: Web Scraping and Named Entity Recognition (NER)
import spacy
import requests
from bs4 import BeautifulSoup
from spacy import displacy

# 1. Initialize the spaCy model
nlp = spacy.load('en_core_web_md')

# 2. Fetch the target webpage from the tutorial instructions
url = "https://www.theguardian.com/us-news/2022/jun/18/donald-trump-election-defense-fundraising-defrauded-fans"
print("=" * 70)
print("TASK 4a: FETCHING & PARSING WEB ARTICLE")
print("=" * 70)
print(f"Target URL: {url}")

page = requests.get(url)

# Use BeautifulSoup to parse the HTML and extract clean text from the paragraphs
soup = BeautifulSoup(page.text, 'html.parser')
article_text = " ".join([p.get_text() for p in soup.find_all('p')])
print(f"Status: Successfully scraped and extracted {len(article_text)} characters of text.\n")

# Process the scraped text through the spaCy pipeline
doc = nlp(article_text)

# 3. Extract and print the named entities in a structured table
print("=" * 70)
print("TASK 4b & 4c: NAMED ENTITY RECOGNITION (NER) EXTRACTION")
print("=" * 70)
print(f"{'NAMED ENTITY':<35} | {'ENTITY CATEGORY'}")
print("-" * 70)
for ent in doc.ents:
    print(f"{ent.text:<35} | {ent.label_}")

# 4. Visualize a manageable span of the document using displaCy
print("\n" + "=" * 70)
print("TASK 4d: DISPLACY ENTITY VISUALIZATION (First 150 Tokens)")
print("=" * 70)
print("Rendering interactive inline visualization...")

# Slice the first 150 tokens so the notebook render remains clean and readable
doc_span = doc[0:150]
displacy.render(doc_span, style='ent', jupyter=True)
print("=" * 70)
```
