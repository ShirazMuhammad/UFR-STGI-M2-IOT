```python
# ==========================================
# CELL 1: LIBRARY IMPORTS & MODEL LOADING
# ==========================================
import spacy
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Load the spaCy English model once at the beginning
# We use this to preprocess text before feeding it into scikit-learn models
nlp = spacy.load("en_core_web_md")

print("Environment ready: spaCy, pandas, and scikit-learn loaded successfully.")
```

```python
# ==========================================
# CELL 2: N-GRAM VOCABULARY GENERATION
# ==========================================
# The target sentence provided in the lab instructions
sentence = ["The Internet of Things master's degree requires natural language processing skills"]

# 5a: Generate a default Unigram vocabulary (1 word per token)
v1 = CountVectorizer()
v1.fit(sentence)
print("=== 5a: UNIGRAM VOCABULARY (1, 1) ===")
print(v1.vocabulary_)

# 5b: Generate Unigrams + Bigrams (1 to 2 words per token)
v2 = CountVectorizer(ngram_range=(1, 2))
v2.fit(sentence)
print("\n=== 5b: N-GRAM RANGE (1, 2) VOCABULARY ===")
print(v2.vocabulary_)

# 5c: Generate Unigrams + Bigrams + Trigrams (1 to 3 words per token)
v3 = CountVectorizer(ngram_range=(1, 3))
v3.fit(sentence)
print("\n=== 5c: N-GRAM RANGE (1, 3) VOCABULARY ===")
print(v3.vocabulary_)
```

```python
# ==========================================
# CELL 3: PIPELINE - SPACY TO SCIKIT-LEARN
# ==========================================
# 5d: Define a function to clean text using spaCy before vectorization
def preprocess_text(text):
    """
    Parses text with spaCy, removes stop words and punctuation,
    and returns a clean string of lemmatized words.
    """
    doc = nlp(text)
    filtered_tokens = [
        token.lemma_ for token in doc
        if not token.is_stop and not token.is_punct
    ]
    return " ".join(filtered_tokens)

# 5d: Define the initial corpus
corpus = [
    "The Internet of Things master's ",
    "master's degree requires",
    "natural language processing skills"
]

# 5e: Clean the entire corpus using our custom spaCy function
preprocessed_corpus = [preprocess_text(doc) for doc in corpus]

# 5f & 5g: Build the vocabulary and transform the text into numerical vectors
vectorizer = CountVectorizer(ngram_range=(1, 2))
corpus_vectors = vectorizer.fit_transform(preprocessed_corpus)

print("=== 5f: CORPUS VOCABULARY MAPPING ===")
print(vectorizer.vocabulary_)

print("\n=== 5g: TRANSFORMED CORPUS VECTORS ===")
print(corpus_vectors.toarray())

# 5h: Test the pipeline on a brand new, unseen sentence
test_sentence = ["master's degree STGI"]
preprocessed_test = [preprocess_text(test_sentence[0])]

# Notice we use .transform() here, NOT .fit_transform(), to use the existing vocabulary
test_vector = vectorizer.transform(preprocessed_test)

print("\n=== 5h: TEST SENTENCE TRANSFORMATION ===")
print(f"Original:     {test_sentence[0]}")
print(f"Preprocessed: {preprocessed_test[0]}")
print(f"Vector:       {test_vector.toarray()}")
```

```python
# 1. TF-IDF Example
# ==========================================
# EXERCISE 1a & 1b: IMPORT, FIT, AND VECTORIZE
# ==========================================
# 1a: Import the TfidfVectorizer (and pandas for clean formatting later)
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# 1b: Define the raw text corpus provided in the assignment
tfidf_corpus = [
    "Thor eating pizza, Loki is eating pizza, Ironman ate pizza already",
    "Apple is announcing new iphone tomorrow",
    "Tesla is announcing new model-3 tomorrow",
    "Google is announcing new pixel-6 tomorrow",
    "Microsoft is announcing new surface tomorrow",
    "Amazon is announcing new eco-dot tomorrow",
    "I am eating biryani and you are eating grapes"
]

# Initialize the TF-IDF Vectorizer
tfidf = TfidfVectorizer()

# Fit the model (learn the vocabulary) and transform the text into vectors simultaneously
tfidf_matrix = tfidf.fit_transform(tfidf_corpus)
print("=== TF-IDF Example (Part a and b) ===")
print("Status: TfidfVectorizer successfully imported, fitted, and applied to the corpus.")
```

```python
# ==========================================
# EXERCISE 1c: PRINT THE TF-IDF VOCABULARY
# ==========================================
print("=== 1c: TF-IDF VOCABULARY MAPPING ===")

# .vocabulary_ returns a dictionary where the key is the word
# and the value is its assigned column index in the matrix
vocab = tfidf.vocabulary_
print(vocab)
```

```python
# ==========================================
# EXERCISE 1d: PRINT INVERSE DOCUMENT FREQUENCY (IDF)
# ==========================================
print("=== 1d: INVERSE DOCUMENT FREQUENCY (IDF) SCORES ===")

# Extract feature names (the words) and their corresponding IDF scores
words = tfidf.get_feature_names_out()
idf_scores = tfidf.idf_

# Map the words to their scores using a pandas DataFrame for professional formatting
idf_df = pd.DataFrame({
    "Word": words,
    "IDF Score": idf_scores
}).sort_values(by="IDF Score", ascending=False) # Sort to show the rarest words first

# Print without the index column for a cleaner look
print(idf_df.to_string(index=False))
```

```python
# ==========================================
# EXERCISE 1e: PRINT TRANSFORMED OUTPUTS
# ==========================================
print("=== 1e: TRANSFORMED TF-IDF MATRIX ===")

# Convert the sparse matrix into a dense NumPy array so we can view it
dense_tfidf = tfidf_matrix.toarray()

# Load it into a DataFrame, using the extracted words as the column headers
tfidf_df = pd.DataFrame(dense_tfidf, columns=tfidf.get_feature_names_out())

# Round to 3 decimal places to match your provided output and keep it readable
print(tfidf_df.round(3))
```
