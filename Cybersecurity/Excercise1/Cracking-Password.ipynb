```python
import hashlib
import itertools
import string
import time

def hash_password(password):
    """Converts a plaintext string into a secure SHA-256 hash."""
    return hashlib.sha256(password.encode('utf-8')).hexdigest()

# 1. Setup: Create a simple database with usernames and hashed passwords.
# 'abc' is a short password for brute forcing.
# 'baseball' is pulled directly from the 10k-most-common-passwords.txt file
database = {
    "alice": hash_password("zzzz"),
    "bob": hash_password("marine")
}

# 2. Task: Write a Python function to brute force a password.
def brute_force_attack(target_hash, max_length=4):
    """Tries every possible combination of lowercase letters until the hash matches."""
    print("[*] Starting Brute Force Attack...")
    charset = string.ascii_lowercase

    # Generate combinations from length 1 up to max_length
    for length in range(1, max_length + 1):
        for guess_tuple in itertools.product(charset, repeat=length):
            # Join the tuple of characters into a single string (e.g., ('a', 'b') -> 'ab')
            guess = ''.join(guess_tuple)

            # Hash the guess and compare it to the target
            if hash_password(guess) == target_hash:
                return guess
    return None

# 3. Task: Write a Python function to perform a dictionary attack.
def dictionary_attack(target_hash, filename="10k-most-common-passwords.txt"):
    """Reads common passwords from a file and hashes them to find a match[cite: 2]."""
    print(f"[*] Starting Dictionary Attack using '{filename}'...")
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            for line in file:
                guess = line.strip() # Removes invisible line breaks

                if hash_password(guess) == target_hash:
                    return guess
    except FileNotFoundError:
        return "Error: Dictionary file not found. Ensure the txt file is in the same folder."
    return None

# 4. Task: Crack the password for specific users by using both methods
if __name__ == "__main__":

    # Crack Alice's password using Brute Force
    print(f"--- Cracking Alice's Password (Hash: {database['alice'][:10]}...) ---")
    start_time = time.time()
    brute_result = brute_force_attack(database["alice"], max_length=4)

    if brute_result:
        print(f"[+] Success! Password is: '{brute_result}' (Found in {time.time() - start_time:.2f} seconds)\n")
    else:
        print("[-] Failed to brute force password.\n")

    # Crack Bob's password using the Dictionary Attack
    print(f"--- Cracking Bob's Password (Hash: {database['bob'][:10]}...) ---")
    start_time = time.time()
    dict_result = dictionary_attack(database["bob"], "10k-most-common-passwords.txt")

    if dict_result:
        print(f"[+] Success! Password is: '{dict_result}' (Found in {time.time() - start_time:.4f} seconds)\n")
    else:
        print("[-] Failed to find password in dictionary.\n")
```
