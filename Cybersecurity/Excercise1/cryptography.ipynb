```python
import math


# EXERCISE 1: Understanding the Modulus

print("=== EXERCISE 1 ===")
P, k, m = 20, 7, 26
C = (P + k) % m
print(f"Plaintext P = {P}, Key k = {k}, Modulus m = {m}")
print(f"Ciphertext C = ({P} + {k}) % {m} = {C}\n")



# EXERCISE 2: Modular Addition Encryption

print("=== EXERCISE 2 ===")
m, k = 26, 7
plaintexts = [0, 1, 5, 10, 19, 20, 25]
print(f"Encrypting plaintexts {plaintexts} with k = {k}, m = {m}:")
for P in plaintexts:
    C = (P + k) % m
    print(f"  P = {P:2d}  -->  C = {C:2d}")
print()



# EXERCISE 3: Discovering Decryption

print("=== EXERCISE 3 ===")
def encrypt_add(P, k, m):
    return (P + k) % m

def decrypt_add(C, k, m):
    return (C - k) % m

m, k, C = 26, 7, 1
P_recovered = decrypt_add(C, k, m)
C_reencrypted = encrypt_add(P_recovered, k, m)

print(f"Ciphertext C = {C}")
print(f"Recovered Plaintext P = decrypt_add({C}, {k}, {m}) = {P_recovered}")
print(f"Re-encrypted C = encrypt_add({P_recovered}, {k}, {m}) = {C_reencrypted}\n")



# EXERCISE 4: Modular Multiplication

print("=== EXERCISE 4 ===")
P, k, m = 20, 5, 26
C = (P * k) % m
print(f"Multiplication Encryption: P = {P}, k = {k}, m = {m} --> C = {C}")

print("Evaluating all P from 1 to 25:")
mult_results = {p: (p * k) % m for p in range(1, 26)}
print(f"{mult_results}\n")



# EXERCISE 5: How Do We Divide Modulo m?

print("=== EXERCISE 5 ===")
k, m = 5, 26
k_inverse = None

# Testing values to find 5*x === 1 mod 26
for x in range(m):
    if (k * x) % m == 1:
        k_inverse = x
        print(f"The inverse of {k} mod {m} is: {x}")
        break

C = 22
P_recovered = (C * k_inverse) % m
print(f"Decrypting C = {C} using inverse {k_inverse}: P = ({C} * {k_inverse}) % {m} = {P_recovered}\n")



# EXERCISE 6: Understanding the Multiplicative Inverse

print("=== EXERCISE 6 ===")
pairs = [(5, 26), (7, 26), (2, 26), (13, 26), (3, 26), (5, 25)]
print("Checking Table 1 Pairs:")
for k_val, m_val in pairs:
    gcd_val = math.gcd(k_val, m_val)
    exists = "Yes" if gcd_val == 1 else "No"
    inv_str = f"{pow(k_val, -1, m_val)}" if exists == "Yes" else "None"
    print(f"  k = {k_val:2d}, m = {m_val:2d} | gcd = {gcd_val:2d} | Inverse Exists? {exists:3s} | Inverse = {inv_str}")
print()



# EXERCISE 7: Why Some Multiplication Keys Cannot Decrypt

print("=== EXERCISE 7 ===")
m, k = 26, 2
print(f"Testing invalid key k = {k} mod {m} for P = 1..13:")
for p in range(1, 14):
    c = (p * k) % m
    print(f"  P = {p:2d}  -->  C = {c:2d}")

e1 = (1 * k) % m
e14 = (14 * k) % m
print(f"Collision Check: E(1) = {e1}, E(14) = {e14} (Both plaintexts produce the same ciphertext!)\n")



# EXERCISE 8: Build a Complete Multiplication Cipher

print("=== EXERCISE 8 ===")
def encrypt_mult(P, k, m):
    return (P * k) % m

def decrypt_mult(C, k_inverse, m):
    return (C * k_inverse) % m

m, k, P = 26, 5, 20
k_inv = pow(k, -1, m)
C = encrypt_mult(P, k, m)
P_rec = decrypt_mult(C, k_inv, m)

print(f"P = {P} --> Encrypted C = {C} --> Decrypted P_recovered = {P_rec}\n")



# EXERCISE 9: Key Generation

print("=== EXERCISE 9 ===")
m = 26
valid_keys = [k_val for k_val in range(1, m) if math.gcd(k_val, m) == 1]
print(f"Valid keys for m = {m} (Total: {len(valid_keys)}): {valid_keys}")

print("Key-Inverse Pairs:")
for k_val in valid_keys:
    inv = pow(k_val, -1, m)
    print(f"  Key k = {k_val:2d}  -->  Inverse k^-1 = {inv:2d}")
print()



# EXERCISE 10: Sender and Receiver Scenario

print("=== EXERCISE 10 ===")
m, k_alice, P_alice = 26, 7, 19
k_bob_inv = pow(k_alice, -1, m)

C_sent = (P_alice * k_alice) % m
P_bob_received = (C_sent * k_bob_inv) % m

print(f"Alice sends encrypted message C = ({P_alice} * {k_alice}) % {m} = {C_sent}")
print(f"Bob decrypts message P = ({C_sent} * {k_bob_inv}) % {m} = {P_bob_received}\n")

# EXERCISE 11: What Information Must Be Shared?

print("=== EXERCISE 11 ===")
print("Parameters: Modulus m = 26, Key k = 5")
print("1. Public Information : Modulus (m = 26) and algorithm structure.")
print("2. Secret Information : Secret key (k = 5).")
print("3. Inverse Calculation: Yes, receiver calculates k^-1 = 21 using gcd/Extended Euclidean.")
print("4. Transmission      : Sender does NOT transmit k^-1; receiver computes it locally.\n")


# EXERCISE 12: Comparing Addition and Multiplication Ciphers

print("=== EXERCISE 12 ===")
matrix = [
    ("Property", "Scheme A (Addition)", "Scheme B (Multiplication)"),
    ("Encryption Formula", "C = (P + 7) mod 26", "C = (P * 7) mod 26"),
    ("Decryption Formula", "P = (C - 7) mod 26", "P = (C * 15) mod 26"),
    ("Inverse Operation", "Additive Inverse (-k)", "Multiplicative Inverse (k^-1)"),
    ("Every Key Reversible?", "Yes", "No"),
    ("Reversibility Condition", "None (Always valid)", "gcd(k, m) == 1")
]
print(f"{matrix[0][0]:<25} | {matrix[0][1]:<22} | {matrix[0][2]:<30}")
print("-" * 82)
for row in matrix[1:]:
    print(f"{row[0]:<25} | {row[1]:<22} | {row[2]:<30}")
print()

# EXERCISE 13: Introduction to Modular Exponentiation

print("=== EXERCISE 13 ===")
P, k, m = 3, 4, 26
C = pow(P, k, m)
print(f"Modular Exponentiation: {P}^{k} mod {m} = {C}")

exp_results = {p: pow(p, k, m) for p in range(1, 26)}
print(f"Outputs for P in 1..25 with k = 4:\n{exp_results}\n")


# EXERCISE 14: Comprehensive Final Challenge

print("=== EXERCISE 14 ===")
def run_final_challenge(m=29, k=12, P=17):
    if math.gcd(k, m) != 1:
        print(f"Error: gcd({k}, {m}) != 1. Key is invalid!")
        return

    k_inv = pow(k, -1, m)
    C = (P * k) % m
    P_rec = (C * k_inv) % m

    print(f"Parameters: Modulus m = {m}, Key k = {k}, Plaintext P = {P}")
    print(f"  1. Multiplicative Inverse k^-1 mod {m} = {k_inv}")
    print(f"  2. Ciphertext C = ({P} * {k}) % {m} = {C}")
    print(f"  3. Recovered Plaintext P = ({C} * {k_inv}) % {m} = {P_rec}")
    assert P == P_rec, "Decryption check failed!"
    print("  Verification successful: P_original == P_recovered")

run_final_challenge()
```
